```python
import pandas as pd
import glob
import os
```

```python
# Handling original csv dataset
original_csvs = ['../DataSet/raw/Phishing_Email.csv', '../DataSet/raw/Phishing_validation_emails.csv']

dfs_original = []
for file in original_csvs:
    df = pd.read_csv(file)

    #confirm column exist
    df = df[['Email Text', 'Email Type']].copy()

    #label clearing
    df['Email Type'] = df['Email Type'].astype(str).str.strip()

    #remove unuseful data
    df = df[df['Email Text'].str.len() > 30]
    df.drop_duplicates(subset = ['Email Text'], inplace = True)
    
    dfs_original.append(df)
    print(f"Ori. CSV {os.path.basename(file)}: {len(df)}")

    #counting number of phishing in ori. dataset
    phishing_count = (df['Email Type'] == 'Phishing Email').sum()
    safe_count = (df['Email Type'] == 'Safe Email').sum()
    other_count = len(df) - phishing_count - safe_count

    print(f" |- Phishing: {phishing_count} ")
    print(f" |- Safe: {safe_count} ")
    print(f" |- Other: {other_count} ")
```

```python
#Handling mbox dataset (Data has been cleaned with transfer from mbox to csv)
mbox_clean_csv = '../DataSet/raw/combined_mbox_phishing_clean_2text.csv'

df_mbox = pd.read_csv(mbox_clean_csv)
df_mbox['Email Type'] = df_mbox['Email Type'].astype(str).str.strip()

print(f"mbox Total: {len(df_mbox)}")
```

```python
#Combine all files

all_dfs = dfs_original + [df_mbox]
final_df = pd.concat(all_dfs, ignore_index = True)

#remove duplicate between all files
final_df.drop_duplicates(subset = ['Email Text'], inplace = True)
print(f"final case number: {len(final_df)}")

final_phishing = (final_df['Email Type'] == 'Phishing Email').sum()
final_safe = (final_df['Email Type'] == 'Safe Email').sum()
print(f"Phishing mail: {final_phishing}")
print(f"Phishing ratio: {final_phishing / len(final_df):.2%}")
```

```python
#output file

final_output = 'final_complete_dataset.csv'
final_df.to_csv(final_output, index=False, encoding='utf-8')

final_df.head()
```
