```python
import pandas as pd
```

```python
df = pd.read_csv('../DataSet/raw/final_complete_dataset.csv')

#taken out the phishing email
phishing_df = df[df['Email Type'] == 'Phishing Email'].copy()

#random 2000
sample_size = min(2000, len(phishing_df))
label_df = phishing_df.sample(n = sample_size, random_state=42).reset_index(drop = True)

#adding labels columns
triggers = ['Urgency', 'Scarcity','Authority', 'Fear', 'Social Proof', 'Reciprocity', 'Liking']
for trigger in triggers:
    label_df[trigger] = 0

label_df.to_csv('phishing_for_labeling_2000.csv', index=False)
label_df.head()
```
